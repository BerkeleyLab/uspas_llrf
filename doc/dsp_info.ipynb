# LLRF DSP application information

DSP calibration factors and initialization registers are generated based on `settings.json`.

This document shows the values for each application.

```python
from uspas_llrf import LLRFShell, dsp_config
from pprint import pprint
```

## USPAS

```python
dsp_config['USPAS']
```

```python
llrf_uspas = LLRFShell(dsp_config['USPAS'], wave_samp_per=1)
pprint(llrf_uspas.cal_factors)
pprint(llrf_uspas.init_regs)
```
